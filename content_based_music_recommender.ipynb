### Import Libraries

```python
import pandas as pd
import joblib
from sklearn.feature_extraction.text import TfidfVectorizer
import re
```

### Load Dataset

```python
songs=pd.read_csv('music.csv')
```

### Drop 'Track_id' Column

```python
songs = songs.drop(columns=['Track_id'])
```

### Define Text Cleaning Function

```python
def clean_text(s):
    import re
    return re.sub(r'\s+', ' ', re.sub(r'[^A-Za-z0-9\' ]+', ' ', str(s))).strip()
```

### Create 'content' Column

```python
songs['content'] = (songs['Song'] + ' ' + songs['Artist'] + ' ' + songs['Genre'] + ' ' + songs['Lyrics'] + ' ' + songs['Song year'].astype(str)).apply(clean_text)
```

### TF-IDF Vectorization

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(stop_words='english', max_features=50000, ngram_range=(1,2))
tfidf_matrix = tfidf.fit_transform(songs['content'])

print("TF-IDF trained! Matrix shape:", tfidf_matrix.shape)
```

### Save Model and Data

```python
joblib.dump(tfidf, "tfidf_model.joblib")
joblib.dump(tfidf_matrix, "tfidf_matrix.joblib")
songs.to_csv("songs.csv", index=False)

print("Model and data saved successfully!")
```
